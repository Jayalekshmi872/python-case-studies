```python
total_revenue = 0
for sale in sales_data:
  total_revenue += sale['quantity']* sale['price_per_item']
print('Total revenue (INR):', total_revenue)
```

# 3.1

```python
def apply_discount(orders, discount_rate):
    discounted_orders = []
    for order in orders:
        # Create a new dictionary for each order so original is not modified
        new_order = order.copy()
        new_order["price_per_item"] = new_order["price_per_item"] * (1 - discount_rate)
        discounted_orders.append(new_order)
    return discounted_orders

# Apply 10% discount
discounted_sales = apply_discount(sales_data, 0.10)

# Print original dataset (unchanged)
print("Original sales_data:")
for sale in sales_data:
    print(sale)

# Print discounted dataset
print("\nDiscounted sales_data:")
for sale in discounted_sales:
    print(sale)
```
